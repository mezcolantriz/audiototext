```python
from faster_whisper import WhisperModel

# Modelo
model = WhisperModel("small")

# Transcribir
segments, info = model.transcribe(
    "clase.wav",
    language="es"
)

texto = ""

for segment in segments:
    texto += segment.text + " "

print(texto)

with open("apuntes.txt", "w", encoding="utf-8") as f:
    f.write(texto)
```
